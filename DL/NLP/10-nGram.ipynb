```python
text = ["this movie was not good"]
##use bigrams

# cv = CountVectorizer(ngram_range=(2,2))
cv = CountVectorizer(ngram_range=(2,3))
print(cv.fit_transform(text).toarray())

print(cv.vocabulary_)
print(cv.get_feature_names_out())
```
