```python
##tokenization --- breaking down a big thing into smaller parts
##paragrapgh ---> sentences
##sentence---> words
```

```python
import nltk
corpus = """following his decision to relocate the Mughal's capital from Agra to Delhi. Originally adorned in red and white, the fort's design is attributed to Ustad Ahmad Lahori,
the architect behind the Taj Mahal."""
```

```python
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
```

```python
doc = sent_tokenize(corpus)
for i in doc:
    print(i)
    print("\n")
    print(word_tokenize(i))
    print("\n")
    print("\n")
    print("\n")
```

```python
from nltk.tokenize import wordpunct_tokenize
##Punactuation will be removed -- . , ? ! : : 
doc = wordpunct_tokenize(corpus)
# print(doc)
doc
```

```python
##tree band word tokenizer  -- more accurate 

from nltk.tokenize import TreebankWordTokenizer
TBT = TreebankWordTokenizer()
doc = TBT.tokenize(corpus)
doc
```
