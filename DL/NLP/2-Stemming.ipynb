```python
##stemming -- reducing words to their root form
#connect , connected , connecting--->connect
##plays , played , playing --> play 
##runs , run , running --> run
##happens , happen , happening --> happen
```

```python

words = ["eating","eats","eaten","writing","writes","programming",
         "programms","history","finally","finalized"]
```

```python
##porter stemmer -- most popular stemmer
##note -- output word may not always be a valid english word 
#pip install nltk -- 
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()

for word in words:
    print(word,"------------->",ps.stem(word))
```

```python
ps.stem("sitting")
```

```python
##snowball stemmer -- updated version porter stemmer //porter2 

from nltk.stem import SnowballStemmer
ss = SnowballStemmer('english')

for word in words:
    print(word,"------------->",ss.stem(word))
```

```python
##regex stemmer 
#regex -- regular expression -- pattern matching--string 
from nltk.stem import RegexpStemmer

rs = RegexpStemmer('ing$|s$|e$|able$|ible$')

for word in words:
    print(word,"------------->",rs.stem(word))

```
